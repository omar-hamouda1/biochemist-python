```python
# Michaelis-Menten Kinetics — First Calculation
Km = 15.0                # Km in micromolar (µM)
Vmax = 100.0             # Vmax in nanomoles/sec
substrate_concentration = 8.0   # Substrate concentration in µM

velocity = (Vmax * substrate_concentration) / (Km + substrate_concentration)
print(velocity)
```

```python
# Substrate concentrations for a Michaelis-Menten experiment (in µM)
substrate_concs = [1.0, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0, 15.0, 20.0, 30.0, 50.0, 75.0, 100.0]

# Find out how many concentrations we have
s_length = len(substrate_concs)

# Print a nice message
print('This list contains', s_length, 'substrate concentrations')
```

```python
# Access the first element (index 0)
print("First concentration:", substrate_concs[0])

# Access the sixth element (index 5) — which is 8.0
print("Sixth concentration:", substrate_concs[5])

# Access the last element (index 12)
print("Last concentration:", substrate_concs[12])
```

```python
# Slice 1: First 3 concentrations (elements 0, 1, 2)
first_three = substrate_concs[0:3]
print("First three:", first_three)

# Slice 2: Same as above, but shorter syntax
first_three_short = substrate_concs[:3]
print("First three (short):", first_three_short)

# Slice 3: Last 5 concentrations (from element 8 to the end)
last_five = substrate_concs[8:]
print("Last five:", last_five)
```

```python
# Make sure we have our values ready
Km = 15.0
Vmax = 100.0
substrate_concs = [1.0, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0, 15.0, 20.0, 30.0, 50.0, 75.0, 100.0]

# Loop over each substrate concentration and calculate velocity
for number in substrate_concs:
    velocity = (Vmax * number) / (Km + number)
    print(velocity)
```

```python
# Step 1: Define an empty list OUTSIDE the loop
velocities = []

# Step 2: Define our values
Km = 15.0
Vmax = 100.0
substrate_concs = [1.0, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0, 15.0, 20.0, 30.0, 50.0, 75.0, 100.0]

# Step 3: Loop and store each velocity
for number in substrate_concs:
    V = (Vmax * number) / (Km + number)
    velocities.append(V)     # Append to the list

# Step 4: Print the full list
print(velocities)
```

```python
# Reset our values
Km = 15.0
Vmax = 100.0
substrate_concs = [1.0, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0, 15.0, 20.0, 30.0, 50.0, 75.0, 100.0]

# Empty list to store the linear-region velocities
linear_velocities = []

# Loop with a condition
for number in substrate_concs:
    if number < Km/4:          # Only if concentration is below 3.75 µM
        V = (Vmax * number) / (Km + number)
        linear_velocities.append(V)

print(linear_velocities)
```

```python
# Reset
Km = 15.0
Vmax = 100.0
substrate_concs = [1.0, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0, 15.0, 20.0, 30.0, 50.0, 75.0, 100.0]

mid_range = []
for number in substrate_concs:
    if number >= 2.0 and number <= 8.0:   # بين 2 و 8
        V = (Vmax * number) / (Km + number)
        mid_range.append(V)

print(mid_range)
```
