```python
# Import the libraries we need
import os      # For file paths
import glob    # For finding files by pattern
```

### Finding Files with `glob.glob()`

The `glob` library lets us find all files matching a pattern
without writing their names manually. The `*` symbol is a
**wildcard** that matches any sequence of characters.

For example, `*.pdb` matches every file ending in `.pdb`.

```python
# Step 1: Build the search pattern
file_location = os.path.join('data', 'PDB_files', '*.pdb')
print(file_location)

# Step 2: Find all matching files
filenames = glob.glob(file_location)
print(filenames)
```

### Processing Multiple Files with Nested Loops

To process all PDB files, we use a **nested loop** — an outer
loop over the files, and an inner loop over each file's lines.

In each file, we search for the `RESOLUTION.` keyword and extract
the value that follows it.

```python
# Loop through all files, and print the PDB ID alongside the resolution
for f in filenames:
    # Get the PDB ID
    file_name = os.path.basename(f)
    split_filename = file_name.split('.')
    pdb_id = split_filename[0]

    # Read the file
    with open(f, "r") as outfile:
        data = outfile.readlines()

    # Search for the resolution
    for line in data:
        if 'RESOLUTION.' in line:
            res_line = line
            words = res_line.split()
            resolution = float(words[3])
            print(f"{pdb_id}: {resolution} Angstroms")
```

### Writing Results to a File

Printing to the screen is fine for exploration, but for sharing
results, we want to save them to a file. We use `with open('file.txt', 'w')`
to open a file for **writing**.

**Key differences from `print`:**
- Use `.write()` instead of `print()`.
- Python can only write **strings** to files.
- Add `\n` at the end of each line to start a new line.

We use **f-strings** for formatting, and `\t` (tab) to align
the columns.

```python
# Write the results to a new file
with open('resolutions.txt', 'w') as datafile:
    for f in filenames:
        # Get the PDB ID
        file_name = os.path.basename(f)
        split_filename = file_name.split('.')
        pdb_id = split_filename[0]

        # Read the file
        with open(f, "r") as outfile:
            data = outfile.readlines()

        # Search for the resolution
        for line in data:
            if 'RESOLUTION.' in line:
                res_line = line
                words = res_line.split()
                resolution = float(words[3])
                datafile.write(f"{pdb_id}\t{resolution}\n")
```

### Verifying the Output File

To verify that our file was written correctly, we can read it
back using the same `with open()` pattern. This is good practice —
always verify your output.

```python
# Verify the file was created and check its contents
with open('resolutions.txt', 'r') as datafile:
    content = datafile.read()

print(content)
```

### A Note on F-String Formatting

F-strings are not limited to just variables — you can embed
entire sentences, add units, and mix text with values. This is
especially useful when writing human-readable output to a file.

```python
# Demonstration: richer output with f-strings
pdb_id = "4eyr"
resolution = 1.8

# Simple format
print(f"{pdb_id}: {resolution} Angstroms")

# Richer format — full sentences
print(f"For the PDB ID {pdb_id}, the resolution is {resolution} Angstroms.")
```

## 🧬 Project: Extracting the Enzyme Commission (EC) Class

### 🎯 Goal
Extract the **EC class** for each enzyme structure in `data/PDB_files`
and write the results to a new file named `EC_class.txt`.

### 📖 What is an EC Class?
The EC (Enzyme Commission) number is a hierarchical classification
of enzyme function — four numbers separated by dots (e.g., `1.2.1.13`).

### 🛠️ Steps
1. Search for the line containing `EC:` in each PDB file.
2. Split the line to extract the EC number.
3. Write the PDB ID and EC number to `EC_class.txt`.

### 📌 Expected Output
7tim 5.3.1.1

6zt7 3.2.1.55

5eu9 4.2.1.11

3iva 2.1.1.13

2pkr 1.2.1.13

3vnd 4.2.1.20

5veu 1.14.14.1

```python
# Test the search on a single file first
with open(filenames[0], "r") as outfile:
    data = outfile.readlines()

for line in data:
    if 'EC:' in line:
        print(line)
```

### Extracting the EC Number

The EC line contains the keyword `EC:` followed by the classification
number. We use `split()` with a colon as the delimiter to separate
the label from the value, then take the second element.

```python
# Extract the EC number from the test line
ec_line = "COMPND   5 EC: 1.4.3.3"

# Split on colon, take second element, clean whitespace
ec_number = ec_line.split(':')[1].strip()
print(ec_number)
```

### Full Project: Extracting EC Class for All Files

Now we combine everything we've learned:
- `glob` to find files.
- Nested loops to process each file.
- `os.path.basename()` to extract the PDB ID.
- `split(':')` to extract the EC number.
- `with open('EC_class.txt', 'w')` to write the results.

```python
# Write the EC class of each enzyme to a new file
with open('EC_class.txt', 'w') as datafile:
    for f in filenames:
        # Get the PDB ID
        file_name = os.path.basename(f)
        split_filename = file_name.split('.')
        pdb_id = split_filename[0]

        # Read the file
        with open(f, "r") as outfile:
            data = outfile.readlines()

        # Search for the EC line
        for line in data:
            if 'EC:' in line:
                ec_number = line.split(':')[1].strip().rstrip(';')
                datafile.write(f"{pdb_id} {ec_number}\n")
```

```python
# Verify the EC_class.txt file was created
with open('EC_class.txt', 'r') as datafile:
    content = datafile.read()

print(content)
```

## ⚠️ Important Notes

### On File Writing
- Python can only write **strings** to files.
- Use `\n` at the end of every line — otherwise, output will be one long line.
- The `'w'` mode **overwrites** the file. Use `'a'` to append.
- Always use `with open(...)` — the file closes automatically.

### On Batch Processing
- `glob.glob()` does **not** guarantee a sorted order.
- Nested loops process **every** file and **every** line — watch your indentation.
- Each file is opened **within** the outer loop, so the file handle
  is fresh for every iteration.

### On Extracting Values
- Different keywords may need different delimiters (`:` vs whitespace).
- Use `rstrip(';')` or `strip()` to clean trailing characters.
- The number of `EC:` lines can vary between structures.
