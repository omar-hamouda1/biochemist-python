### Locating the Data File

### Importing Data with `pd.read_csv()`

Pandas provides `read_csv()` to load CSV files into a **DataFrame** —
a 2D table with named columns (like a spreadsheet, but programmable).

The first row of the CSV is automatically used as the **column headers**.

```python
# Import the CSV file into a pandas DataFrame
thrombin_df = pd.read_csv(thrombin_file)

# Display the DataFrame
thrombin_df
```

### Exploring the DataFrame

Before analyzing any dataset, we should explore its structure:
- What columns does it have?
- What do the first/last rows look like?
- How can we access specific columns or rows?

Pandas provides several convenient methods for this:
- `.head(n)` — first n rows
- `.tail(n)` — last n rows
- `df['column']` — access a single column
- `df.loc[row]` — access by label
- `df.iloc[row, col]` — access by integer position

```python
# Show the first 5 rows (default)
thrombin_df.head()
```

```python
# Show only the first 3 rows
thrombin_df.head(3)
```

```python
# Show the last 5 rows
thrombin_df.tail()
```

### Accessing a Single Column

To access one column, use square brackets with the column name:
`df['column_name']`. The result is a **Series** — a 1D labeled array.

We'll extract the `Resolution` column — the key measurement
that describes the quality of each crystal structure.

```python
# Access the Resolution column
thrombin_df['Resolution']
```

```python
# Access the Ligand name column
thrombin_df['Ligand name']
```

```python
# Select multiple columns (note the DOUBLE brackets)
thrombin_df[['PDB ID', 'Resolution']]
```

### Accessing a Row — `.loc`

`.loc` accesses data by **label** (index number or column name).
It can retrieve:
- A whole row: `df.loc[row]`
- A single cell: `df.loc[row, column]`

```python
# Access the entire row at index 2
thrombin_df.loc[2]
```

```python
# Access the Resolution value at row 2
thrombin_df.loc[2, 'Resolution']
```

```python
# Same result using .iloc (integer position)
thrombin_df.iloc[2, 2]
```

### Sorting Data — `sort_values()`

Pandas can sort rows by any column. By default, it sorts in
**ascending** order. To sort in place (modify the DataFrame itself),
use `inplace=True` — otherwise, the result is returned as a new DataFrame.

```python
# Sort by Resolution (ascending) — but WITHOUT inplace
thrombin_df.sort_values('Resolution')
```

```python
# Check the original — it should NOT be sorted
thrombin_df
```

```python
# Sort in place — modifies the DataFrame permanently
thrombin_df.sort_values('Resolution', inplace=True)

# Verify the change
thrombin_df
```

```python
# Sort by Resolution in descending order (in-place)
thrombin_df.sort_values('Resolution', ascending=False, inplace=True)

# Display the sorted DataFrame
thrombin_df
```

## 🧬  Enzyme Kinetics Data

The file `data/enzyme_kinetics.csv` contains enzyme kinetics data
(substrate concentration vs. velocity), but the rows are **out of order**.

### 🎯 Goal
1. Load the file into a DataFrame.
2. Sort the data by **substrate concentration** (ascending).
3. Find the **velocity** at a substrate concentration of **20 µM**.

```python
# Build the path to the enzyme kinetics file
kinetics_file = os.path.join('data', 'enzyme_kinetics.csv')
print(kinetics_file)

# Verify it exists (best practice)
if os.path.exists(kinetics_file):
    print("File found ✅")
else:
    print("File not found ❌")
```

```python
# Load the enzyme kinetics data into a DataFrame
kinetics_df = pd.read_csv(kinetics_file)

# Display the DataFrame
kinetics_df
```

```python
# Sort the data by substrate concentration (ascending)
kinetics_df.sort_values('Substrate_conc', inplace=True)

# Display the sorted DataFrame
kinetics_df
```

```python
# Find the velocity at substrate concentration = 20 µM
kinetics_df.loc[9]
```

```python
# Robust: filter by condition instead of hardcoded index
result = kinetics_df[kinetics_df['Substrate_conc'] == 20]
result
```

```python
# Extract just the velocity value
velocity_at_20 = kinetics_df[kinetics_df['Substrate_conc'] == 20]['Velocity'].values[0]
print(f"Velocity at 20 µM: {velocity_at_20:.2f}")
```

## ⚠️ Important Notes

### On DataFrames
- A DataFrame is a 2D table with labeled rows and columns.
- `pandas` is built on top of `numpy`, which provides fast numerical operations.
- The standard import alias `pd` is a universal convention — always use it.

### On Accessing Data
- `.loc` uses **labels** (index names, column names).
- `.iloc` uses **integer positions** (like regular Python indexing).
- When possible, prefer `.loc` — it's clearer and safer.

### On Sorting
- `sort_values()` returns a new DataFrame by default.
- Use `inplace=True` to modify the original DataFrame.
- Add `ascending=False` for descending order.

### On Filtering
- `df[df['col'] == value]` is the standard way to filter rows.
- This is safer than hardcoding row indices (`.loc[9]`).
- It works regardless of row order.
