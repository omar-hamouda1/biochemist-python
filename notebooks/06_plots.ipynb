```python
# Standard library
import os                          # File paths

# Third-party scientific libraries
import numpy as np                 # Numerical operations
import pandas as pd                # Tabular data
from scipy import stats            # Linear regression

# Plotting libraries
import matplotlib.pyplot as plt    # Basic plotting
import seaborn as sns              # Statistical plots
```

### Loading Data & Linear Regression (Recap)

We reuse the exact workflow from Lesson 06:
1. Load `protein_assay.csv` into a DataFrame.
2. Extract X (concentration) and Y (absorbance).
3. Perform linear regression with `scipy.stats.linregress`.

```python
# Build the path to the protein assay file
protein_file = os.path.join('data', 'protein_assay.csv')

# Load the data
results_df = pd.read_csv(protein_file)
print("Columns:", results_df.columns.tolist())

# Extract X and Y
xdata = results_df['Protein Concentration (mg/mL)']   # X: known concentrations
ydata = results_df['A595']                            # Y: absorbance readings

# Perform linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(xdata, ydata)

# Display results
print(f"\nSlope = {slope} /mg/mL")
print(f"Intercept = {intercept}")
print(f"R-squared = {r_value**2}")
print(f"P value = {p_value}")
print(f"Standard error = {std_err}")
```

### Creating a Scatter Plot with Matplotlib

`plt.scatter(x, y)` plots each data point as a separate marker.
Before fitting any line, it's good practice to **look at the data**
to check for outliers and confirm the relationship is roughly linear.

We set the figure size with `plt.figure(figsize=(width, height))`.

```python
# Create a scatter plot of the calibration curve
plt.figure(figsize=(10, 5))    # width=10, height=5 inches
plt.scatter(xdata, ydata)      # plot each (x, y) as a point
plt.show()                     # display the figure
```

### Adding the Best-Fit Line

After plotting the data points, we add the **regression line**
using the slope and intercept from `linregress`.

The line is plotted with: y = slope * x + intercept




Now the plot shows both the **raw data** and the **fitted model**.

```python
# Create the figure
plt.figure(figsize=(10, 6))

# Plot the raw data points
plt.scatter(xdata, ydata, label='Data')

# Plot the best-fit line using the regression equation
plt.plot(xdata, slope * xdata + intercept, color='red', label='Best Fit')

plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(xdata, ydata, label='Data')
plt.plot(xdata, slope * xdata + intercept, color='red', label='Best Fit')
plt.legend()    # ← بيظهر المفتاح
plt.show()
```

### Adding Axis Labels & Equation Annotation

A publication-quality plot needs:
- **Axis labels** with units — `plt.xlabel()` / `plt.ylabel()`
- **Equation annotation** — `plt.annotate()` prints text at a given (x, y).

We use **f-strings** with format specifiers (`.4f`) to display the
slope and intercept with 4 decimal places, and **LaTeX** to render
subscripts like `A₅₉₅`.

```python
# Create the figure
plt.figure(figsize=(10, 6))

# Plot the raw data and the best-fit line
plt.scatter(xdata, ydata, label='Data')
plt.plot(xdata, slope * xdata + intercept, color='red', label='Best Fit')

# Add axis labels (with LaTeX for subscript)
plt.xlabel('Protein Concentration (mg/mL)')
plt.ylabel('$A_{595}$')

# Annotate the equation of the line on the plot
plt.annotate(f'y = {slope:.4f} * x + {intercept:.4f}', xy=(0.25, 1.0))

# Display the figure
plt.show()
```

### Saving the Figure — `plt.savefig()`

To save the figure for reports, publications, or sharing, use
`plt.savefig()`. Key arguments:
- **filename** — string with extension (e.g., `.png`, `.pdf`)
- **dpi** — resolution (600 for publication quality)
- **bbox_inches='tight'** — removes extra whitespace around the figure

The file is saved in the **same directory as the notebook**.

```python
# Create the figure
plt.figure(figsize=(10, 6))

# Plot the raw data and best-fit line
plt.scatter(xdata, ydata, label='Data')
plt.plot(xdata, slope * xdata + intercept, color='red', label='Best Fit')

# Add axis labels
plt.xlabel('Protein Concentration (mg/mL)')
plt.ylabel('$A_{595}$')

# Annotate the equation
plt.annotate(f'y = {slope:.4f} * x + {intercept:.4f}', xy=(0.25, 1.0))

# Save the figure as a high-resolution PNG
plt.savefig('Bradford_plot.png', dpi=600, bbox_inches='tight')

# Display the figure
plt.show()
```

### Seaborn — Regression Plot with Confidence Interval

`seaborn` is built on matplotlib and provides high-level statistical
plots. `sns.regplot()` creates a **scatter plot with a regression line
and a 95% confidence interval** — all in one line.

**Why this is powerful:**
- The shaded band shows the **reliable region** of the calibration curve.
- It **visually communicates** that extrapolation beyond the standards
  is not valid.
- One line of code replaces many matplotlib commands.

```python
# Create a regression plot with confidence interval using Seaborn
plt.figure(figsize=(10, 6))
sns.regplot(x=xdata, y=ydata)
plt.show()
```

```python
# Build the path to the microplate assay file
protein_file2 = os.path.join('data', 'protein_assay2.csv')
print(protein_file2)

# Verify it exists
if os.path.exists(protein_file2):
    print("File found ✅")
else:
    print("File not found ❌")
```

```python
# Load the microplate assay data
results_df2 = pd.read_csv(protein_file2)

# Check column names FIRST (lesson from KeyError!)
print("Columns:", results_df2.columns.tolist())

# Display the DataFrame
results_df2
```

```python
# Extract X and Y (using the actual column names from this file)
xdata = results_df2['mg/mL']       # X: concentration
ydata = results_df2['A595']         # Y: absorbance

# Perform linear regression
slope2, intercept2, r_value2, p_value2, std_err2 = stats.linregress(xdata, ydata)

# Display the regression statistics
print(f"Slope = {slope2}")
print(f"Intercept = {intercept2}")
print(f"R-squared = {r_value2**2}")
print(f"P value = {p_value2}")
print(f"Standard error = {std_err2}")
```

## ⚠️ Important Notes

### On Matplotlib
- `plt.figure(figsize=(w, h))` sets the figure size in inches.
- `plt.savefig()` **must be called before** `plt.show()`.
- `dpi=600` is the standard for **publication-quality** output.
- Use `.pdf` for **vector format** — required by most journals.

### On LaTeX in Matplotlib
- Wrap text in `$...$` to render math notation.
- Use `_` for subscripts: `$A_{595}$` → A₅₉₅.
- Common Greek letters: `$\mu$` (μ), `$\alpha$` (α), `$\sigma$` (σ).

### On Seaborn
- `sns.regplot(x, y)` creates scatter + regression + 95% CI in one line.
- Seaborn **depends on** numpy, scipy, pandas, and matplotlib.
- The confidence band shows where the model is **reliable** — and where
  extrapolation becomes risky.

### On Figure Quality

| **Screen / web** | 72–150 | PNG |
| **Presentation** | 150–300 | PNG |
| **Publication (raster)** | 600 | TIFF / PNG |

## 🧬 Biological Context

### Why Calibration Curves Matter
Protein quantification assays (Bradford, BCA, Lowry) rely on
**calibration curves** built from known standards. The linear
regression line converts raw absorbance into protein concentration.
**R² ≥ 0.99** is the industry standard for acceptable calibration.

### Why Confidence Intervals Matter
The shaded band in Seaborn's regplot shows the **95% confidence
interval** — the range where the true line is likely to lie. In a
calibration curve:
- **Narrow CI** → high confidence, high-quality data.
- **Wide CI** → noisy data; results from those regions are less reliable.

### Real-World Implication
Reporting protein concentrations **outside the standards' range**
is scientifically invalid. The plot's CI band **visually reminds us**
of this limit — an important quality-control habit.
