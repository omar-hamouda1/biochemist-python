## 3️⃣ Applying the Power of Python

```python
# Store the query string in a variable
ID = 'pdbid=8gch'

# Use the variable in the view call
scene = icn3dpy.view(q=ID)
scene
```

### 📝 Section Summary

In this section, we:

- **Reused** the same `view()` call for a different structure.
- **Understood** why variables make code more **flexible** and **readable**.

> **Key takeaway:** Storing query parameters in variables is the first step toward **reusable, modular code** — a pattern we'll extend in the next sections.

## 4️⃣ Adding Commands to the View Function

```python
# Change protein representation to cylinder and plate
scene = icn3dpy.view(
    q='pdbid=1mbo',
    command='style proteins cylinder and plate'
)
scene
```

```python
# Change protein AND heme (chemical) representations
scene = icn3dpy.view(
    q='pdbid=1mbo',
    command='style proteins cylinder and plate; style chemicals ball and stick'
)
scene
```

```python
# Only change chemicals — keep protein default
scene = icn3dpy.view(
    q='pdbid=1mbo',
    command='style chemicals ball and stick'
)
scene
```

```python
scene = icn3dpy.view(
    q='pdbid=1mbo',
    command='style chemicals wireframe'
)
scene
```

```python
# Store each command in a separate variable for readability
ID  = 'pdbid=1mbo'
cpl = '; style proteins cylinder and plate'
bns = '; style chemicals ball and stick'

scene = icn3dpy.view(q=ID, command=cpl + bns)
scene
```

```python
# Same commands, different protein (4HHB)
ID  = 'pdbid=4hhb'
cpl = '; style proteins cylinder and plate'
bns = '; style chemicals ball and stick'

scene = icn3dpy.view(q=ID, command=cpl + bns)
scene
```

### 📝 Section Summary

In this section, we:
- **Learned** to pass the `command` parameter to `view()`.
- **Applied** multiple commands separated by semicolons (`;`).
- **Stored** each command in a variable for flexibility.
- **Compared** two proteins (`1MBO` and `4HHB`) with the same style.

> **Key takeaway:** Commands are strings — combining them with `+` and storing them in variables makes your code **modular** and **easy to customize**.

## 5️⃣ Pulling Commands from an iCN3D Web Page

```python
# Paste the commands captured from the iCN3D web page for 1ETS
scene = icn3dpy.view(
    q='mmdbafid=1ETS',
    command='load mmdb 1ets | parameters &mmdbid=1ets&bu=1; '
            'defined sets; select sets chemicals; set background white; '
            'style chemicals ball and stick; set surface wireframe on; '
            'style proteins cylinder and plate; defined sets; '
            'select sets 1ETS; style proteins cylinder and plate; '
            'color secondary structure yellow'
)
scene
```

```python
# Simplify using variables
ID = '8gch'
scene = icn3dpy.view(
    q='mmdbafid=' + ID,
    command='load mmdb ' + ID + ' | parameters &mmdbid=' + ID + '&bu=1; '
            'defined sets; select sets chemicals; set background white; '
            'style chemicals ball and stick; set surface wireframe on; '
            'style proteins cylinder and plate; defined sets; '
            'select sets ' + ID + '; style proteins cylinder and plate; '
            'color secondary structure yellow'
)
scene
```

### 📝 Section Summary

In this section, we:
- **Learned** the iCN3D **web → Jupyter workflow** (File → Share Link → Copy Commands).
- **Applied** a complex command set to `1ETS`.
- **Refactored** the commands using a **variable** for the PDB ID.
- **Reused** the same view for a different structure (`8GCH`).

> **Key takeaway:** Don't type complex commands manually. Build the view on the iCN3D web page, copy the commands, and paste them into Jupyter. Then **wrap the ID in a variable** for maximum reusability.

## 6️⃣ Creating Functions for Reusable Views

```python
def myview1(PDB_ID):
    """
    Generate a predefined iCN3D view for any PDB ID.
    
    Parameters
    ----------
    PDB_ID : str
        The 4-character PDB identifier (e.g., '1ETS').
    
    Returns
    -------
    scene : icn3dpy widget
        The interactive 3D view.
    """
    PDB_ID = str(PDB_ID)
    scene = icn3dpy.view(
        q='mmdbafid=' + PDB_ID,
        command='load mmdb ' + PDB_ID + ' | parameters &mmdbid=' + PDB_ID + '&bu=1; '
                'defined sets; select sets chemicals; set background white; '
                'style chemicals ball and stick; set surface wireframe on; '
                'style proteins cylinder and plate; defined sets; '
                'select sets ' + PDB_ID + '; style proteins cylinder and plate; '
                'color secondary structure yellow'
    )
    return scene
```

```python
# Call the function with a PDB ID
PDB_ID = '1ETS'
myview1(PDB_ID)
```

```python
# Try a different PDB ID with the same function
PDB_ID = '1d66'
myview1(PDB_ID)
```

### 📝 Section Summary

In this section, we:
- **Defined** a reusable function `myview1(PDB_ID)`.
- **Included** a **docstring** with parameters and return values.
- **Called** the function for `1ETS` (Thrombin) and `1d66` (GAL4-DNA).
- **Understood** the difference between **defining** and **calling** a function.

> **Key takeaway:** Functions turn **complex command strings** into **one-line calls**. This is the **final step** in building a scalable visualization workflow.
